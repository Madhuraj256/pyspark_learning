```python
data = [
    (1, "Alice", 30000),
    (2, "Bob", 40000),
    (3, "Charlie", 50000)
]

df = spark.createDataFrame(data, ["id", "name", "salary"])


```

```python
df.write.format("delta").saveAsTable("employees")
```

```python
df.display()
```

```python
new_data = [(4, "David", 60000)]
new_df = spark.createDataFrame(new_data, ["id", "name", "salary"])

new_df.write.format("delta").mode("append").saveAsTable("employees")
```

```python
from delta.tables import DeltaTable

delta_table = DeltaTable.forName(spark, "employees")

delta_table.update(
    condition="id = 1",
    set={"salary": "35000"}
)
```

```python
delta_table.delete("id = 2")
```

```python
updates = [
    (1, "Alice", 70000),   # update
    (5, "Eve", 45000)      # insert
]

updates_df = spark.createDataFrame(updates, ["id", "name", "salary"])

delta_table.alias("t").merge(
    updates_df.alias("s"),
    "t.id = s.id"
).whenMatchedUpdateAll() \
 .whenNotMatchedInsertAll() \
 .execute()


```
